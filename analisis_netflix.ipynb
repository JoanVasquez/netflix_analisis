# Análisis de títulos de Netflix
Notebook ligero: la lógica reutilizable está en `src/netflix_analisis/`. Ejecuta este archivo desde la raíz del proyecto.

```python
from src.netflix_analisis.config import Config
from src.netflix_analisis.repository import CsvNetflixRepository
from src.netflix_analisis.cleaning import NetflixCleaner
from src.netflix_analisis.analysis import NetflixAnalyzer
from src.netflix_analisis.visualization import NetflixPlotter
import matplotlib.pyplot as plt
```

## 1. Carga e inspección

```python
config = Config()
df = CsvNetflixRepository().load(config.csv_path)
display(df.head())
print("Dimensiones:", df.shape)
display(df.isna().sum())
display(df.dtypes)
display(df.describe())
print("Duplicados:", df.duplicated().sum())
```

## 2. Limpieza

```python
df_limpio = NetflixCleaner(config).clean(df)
display(df_limpio.isna().sum())
display(df_limpio[["release_year", "date_added"]].dtypes)
display(df_limpio[["type", "duration", "duration_minutes"]].head(10))
```

## 3. Análisis

```python
analizador = NetflixAnalyzer(df_limpio, config.missing_label)
titulos_2010 = analizador.titles_after(2010)
top_10_paises = analizador.top_countries()
conteo_generos = analizador.genre_counts()
conteo_directores = analizador.director_counts()
director_principal, obras_director = analizador.top_director_works()
resumen_duracion = analizador.movie_duration_by_country()
peliculas_boxplot, paises_validos = analizador.boxplot_data(min_movies=5)
display(titulos_2010[["title", "type", "release_year"]].head())
display(top_10_paises)
display(conteo_generos.head(10))
display(conteo_directores.head(5))
print("Director principal:", director_principal)
display(obras_director)
display(resumen_duracion)
```

## 4. Visualización

```python
graficos = NetflixPlotter()
graficos.countries(top_10_paises)
plt.show()
graficos.genres(conteo_generos)
plt.show()
graficos.movie_durations(peliculas_boxplot, paises_validos)
plt.show()
```
